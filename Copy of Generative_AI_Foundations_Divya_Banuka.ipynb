# 2. LLM Fundamentals

## What is an LLM?

LLM stands for Large Language Model.

An LLM is an artificial intelligence model trained on a very large amount of text data. It learns patterns in language and can generate human-like text.

Examples of tasks performed by LLMs include:

- Answering questions
- Summarizing text
- Translation
- Text generation
- Code generation
- Question answering
- Content creation

## How does an LLM work?

A simplified LLM workflow is:

User Prompt
↓
Tokenization
↓
Transformer Model
↓
Next-token Prediction
↓
Generated Response

The model processes the input and predicts suitable tokens to generate a response.

## What is a prompt?

A prompt is the instruction or question given to an AI model.

### Example

Prompt:

"Explain Artificial Intelligence in simple words."

The LLM processes this prompt and generates a response based on patterns learned during training.

## Key LLM concepts

### 1. Training

The model learns patterns from large datasets during training.

### 2. Parameters

Parameters are numerical values learned during training that help the model generate outputs.

### 3. Inference

Inference is the process of using a trained model to generate an output for a given input.

# 3. Tokens and Context Windows

## What is a Token?

A token is a small unit of text that an AI language model processes.

A token can represent:

- A complete word
- Part of a word
- A punctuation mark
- A space or other text element

For example:

The sentence:

"AI is useful."

is divided into smaller pieces called tokens before being processed by an LLM.

## Why are Tokens Important?

LLMs process text as tokens rather than directly processing complete sentences.

The number of tokens affects:

- Input size
- Output size
- Processing requirements
- Model usage limits

---

## What is a Context Window?

A context window is the maximum amount of information that an AI model can consider during a single interaction.

It includes information such as:

- User instructions
- Previous conversation
- Documents provided to the model
- The model's generated response

### Simple Example

If a model has a context window of 10,000 tokens, the total information processed within that interaction must fit within that limit.

## Token vs Context Window

| Concept | Meaning |
|---|---|
| Token | Small unit of text processed by the model |
| Context Window | Maximum amount of tokens the model can consider |

### Key Point

A larger context window allows a model to work with longer documents or conversations, subject to the model's specific limits.

# 4. Transformer Architecture

## What is a Transformer?

A Transformer is a neural network architecture widely used in modern language models.

Transformers use a mechanism called **attention** to determine which parts of the input are important when processing information.

## Main Components

### 1. Input

The input text is converted into tokens.

### 2. Embeddings

Tokens are represented as numerical vectors so that the neural network can process them.

### 3. Attention

The attention mechanism helps the model understand relationships between different tokens in the input.

### 4. Transformer Layers

Multiple neural network layers process the information and learn complex patterns.

### 5. Output

The model produces probabilities for possible next tokens and generates the final response.

## Simplified Architecture

Text Input
↓
Tokenization
↓
Embeddings
↓
Attention Mechanism
↓
Transformer Layers
↓
Output Probabilities
↓
Generated Text

## Why are Transformers Important?

Transformers have made it possible to build powerful language models that can process and generate text efficiently.

They are a fundamental architecture behind many modern Generative AI systems.

# 6. AI Hallucinations

## What is an AI Hallucination?

An AI hallucination occurs when an AI model generates information that sounds convincing but is incorrect, unsupported, or fabricated.

## Example

A user asks:

"Who invented a fictional programming language called XYZ?"

If the model does not know that XYZ is fictional but generates a confident answer about its supposed inventor, that answer is an example of a hallucination.

## Why do hallucinations happen?

Some possible reasons include:

- The model generates responses based on learned patterns.
- The required information may not be available to the model.
- The prompt may be ambiguous.
- The model may produce a plausible continuation instead of verifying a fact.

## How can hallucinations be reduced?

Useful approaches include:

- Providing clear prompts
- Supplying reliable reference information
- Asking the model to identify uncertainty
- Verifying important information using trusted sources
- Using retrieval or external knowledge systems when appropriate

## Important Point

AI-generated information should not automatically be treated as fact. Important information should be checked against reliable sources.

# 7. AI Bias and Limitations

## What is AI Bias?

AI bias occurs when an AI system produces outputs that systematically favor or disadvantage certain groups, perspectives, or outcomes.

Bias can enter an AI system through:

- Training data
- Data collection methods
- Human decisions
- Model design
- Evaluation methods

## Example

If a model is trained on data that does not adequately represent different groups, its outputs may perform differently across those groups.

## Limitations of Generative AI

Generative AI systems have several limitations:

- They can generate incorrect information.
- They may produce hallucinations.
- Their knowledge may have limitations depending on the system and data available to it.
- They can reflect biases present in training data.
- They may misunderstand ambiguous prompts.
- AI-generated content may require human verification.

## How to Handle These Limitations

Users should:

1. Provide clear instructions.
2. Verify important information.
3. Use reliable sources for factual claims.
4. Avoid sharing sensitive information unnecessarily.
5. Keep a human involved in important decisions.

## Key Point

Generative AI is a useful tool, but its output should be evaluated rather than automatically accepted as correct.
